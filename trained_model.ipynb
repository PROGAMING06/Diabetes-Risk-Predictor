```python
import pandas as pd
from sklearn.model_selection import train_test_split
from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pickle

# --- Load & clean (unchanged from your notebook) ---
df = pd.read_csv("diabetes.csv")
df.dropna(inplace=True)
df.drop_duplicates()

df1 = df.drop(["Patient_ID", "AI_Health_Recommendation", "Doctor_Consultation_Needed",
               "Diabetes_Risk", "BMI_CATEGORY"], axis=1)

categorical_columns = df1.select_dtypes(include="object").columns.tolist()
numeric_columns = df1.select_dtypes(exclude="object").columns.tolist()

df1[categorical_columns] = df1[categorical_columns].fillna("Unknown").astype(str)
df1[numeric_columns] = df1[numeric_columns].fillna(df1[numeric_columns].median())

x = df1.drop(["Diabetes_Risk_Score"], axis=1)
y = df1["Diabetes_Risk_Score"]

# --- Three-way split: train / validation / test ---
# Step 1: carve off the test set first. It is not touched again until final scoring.
x_trainval, x_test, y_trainval, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Step 2: split the remaining 80% into train and validation.
# test_size=0.25 here is 0.25 * 0.8 = 0.2 of the original data,
# so overall this works out to 60% train / 20% val / 20% test.
x_train, x_val, y_train, y_val = train_test_split(
    x_trainval, y_trainval, test_size=0.25, random_state=42
)

model = CatBoostRegressor(
    loss_function="RMSE",
    eval_metric="MAE",
    iterations=1000,
    learning_rate=0.05,
    depth=8,
    l2_leaf_reg=5,
    random_seed=42,
    verbose=100,
    early_stopping_rounds=100,
)

# Validation set drives early stopping — test set stays untouched.
model.fit(
    x_train, y_train,
    cat_features=categorical_columns,
    eval_set=(x_val, y_val),
    verbose=100,
)

# Test set is used exactly once, here, for the final honest score.
predictions = model.predict(x_test)
print(f"MAE: {mean_absolute_error(y_test, predictions):.4f}")
print(f"RMSE: {mean_squared_error(y_test, predictions) ** 0.5:.4f}")
print(f"R²: {r2_score(y_test, predictions):.4f}")

model.save_model("diabetes_risk_score_catboost.cbm")
pickle.dump(model, open("model.pkl", "wb"))
```
